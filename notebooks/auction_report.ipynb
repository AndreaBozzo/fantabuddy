# Fantabuddy — report parametrico
Impostare `FANTABUDDY_DB`, `FANTABUDDY_BUILD_ID` e, facoltativamente, `FANTABUDDY_REPORT`.

```python
import os
from pathlib import Path

from fantabuddy.db import database
from fantabuddy.report import render_report

db = Path(os.getenv("FANTABUDDY_DB", "data/warehouse/fantabuddy.duckdb"))
build_id = os.environ["FANTABUDDY_BUILD_ID"]
output = Path(os.getenv("FANTABUDDY_REPORT", f"outputs/{build_id}/report-notebook.html"))
with database(db) as connection:
    render_report(connection, build_id, output)
output
```
